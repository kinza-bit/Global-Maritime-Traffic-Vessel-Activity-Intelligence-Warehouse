```python
import calendar
import json
import time
from datetime import datetime, timezone
import requests

# ── Configuration ────────────────────────────────────────────────────────────
SA_NAME = "stmaritimeks"
BRONZE_ROOT = f"abfss://bronze@{SA_NAME}.dfs.core.windows.net"
WATERMARK_PATH = f"{BRONZE_ROOT}/_state/watermark.json"
BASE_URL = "https://gateway.api.globalfishingwatch.org/v3"

KV_SCOPE = "maritime-kv"
API_TOKEN = dbutils.secrets.get(scope=KV_SCOPE, key="gfw-api-token")

HEADERS = {
    "Authorization": f"Bearer {API_TOKEN}",
    "Content-Type": "application/json",
}

MAX_ROWS_PER_CHUNK = 15000

WORLD_DATASETS = {
    "gap": "public-global-gaps-events:latest",
    "fishing": "public-global-fishing-events:latest",
    "loitering": "public-global-loitering-events:latest",
    "encounter": "public-global-encounters-events:latest",
}

PORT_VISIT_DATASET = "public-global-port-visits-events:latest"

PORT_BOXES = {
    "shanghai": [121.30, 30.50, 122.30, 31.50],
    "singapore": [103.55, 1.10, 104.10, 1.40],
    "busan": [128.90, 34.95, 129.20, 35.20],
    "hong_kong": [113.80, 22.15, 114.35, 22.45],
    "rotterdam": [3.95, 51.85, 4.55, 52.05],
    "antwerp": [4.15, 51.20, 4.45, 51.40],
    "hamburg": [9.75, 53.45, 10.05, 53.60],
    "la_longbeach": [-118.35, 33.68, -118.10, 33.82],
    "new_york_nj": [-74.22, 40.55, -73.95, 40.75],
    "jebel_ali": [54.95, 24.90, 55.15, 25.08],
    "port_klang": [101.25, 2.90, 101.45, 3.10],
    "santos": [-46.40, -24.05, -46.25, -23.90],
    "karachi": [66.85, 24.72, 67.05, 24.88],
}


def get_next_month_window():
    raw = dbutils.fs.head(WATERMARK_PATH, 10000)
    state = json.loads(raw)
    last_ym = state["last_loaded_month"]  # e.g., "2025-02"
    y, m = map(int, last_ym.split("-"))
    if m == 12:
        ny, nm = y + 1, 1
    else:
        ny, nm = y, m + 1
    next_ym = f"{ny:04d}-{nm:02d}"
    last_day = calendar.monthrange(ny, nm)[1]
    return next_ym, f"{next_ym}-01", f"{next_ym}-{last_day:02d}"


def bbox_to_polygon(bbox):
    min_lon, min_lat, max_lon, max_lat = bbox
    return {
        "type": "Polygon",
        "coordinates": [[
            [min_lon, min_lat],
            [max_lon, min_lat],
            [max_lon, max_lat],
            [min_lon, max_lat],
            [min_lon, min_lat],
        ]],
    }


def fetch_events_with_retry(dataset: str, start_date: str, end_date: str, geometry=None, max_rows=15000):
    url = f"{BASE_URL}/events"
    limit = 1000
    offset = 0
    collected = []
    retries_used = 0

    # GFW /v3/events requires a geometry polygon; default to worldwide if None
    if geometry is None:
        geometry = bbox_to_polygon([-180, -90, 180, 90])

    while len(collected) < max_rows:
        params = {"limit": min(limit, max_rows - len(collected)), "offset": offset}
        payload = {
            "datasets": [dataset],
            "startDate": start_date,
            "endDate": end_date,
            "geometry": geometry,
        }

        for attempt in range(1, 6):
            resp = requests.post(url, headers=HEADERS, params=params, json=payload, timeout=60)
            if resp.status_code in (200, 201):
                break
            if resp.status_code in (429, 500, 502, 503, 504):
                retries_used += 1
                wait_sec = 2 ** attempt
                print(f"  [HTTP {resp.status_code}] Retrying in {wait_sec}s (attempt {attempt}/5)...")
                time.sleep(wait_sec)
            else:
                raise RuntimeError(f"HTTP {resp.status_code}: {resp.text[:300]}")
        else:
            raise RuntimeError(f"Failed after 5 retries for {dataset} {start_date}..{end_date}")

        data = resp.json()
        entries = data.get("entries", [])
        if not entries:
            break

        collected.extend(entries)
        if len(entries) < params["limit"]:
            break
        offset += len(entries)
        time.sleep(0.25)

    return collected, retries_used


def run_incremental_ingestion():
    ym_str, win_start, win_end = get_next_month_window()
    ingest_date = datetime.now(timezone.utc).strftime("%Y-%m-%d")
    run_id = f"incremental_{datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%SZ')}"
    print(f"Starting incremental ingestion for window: {ym_str} ({win_start} to {win_end})")

    manifest_records = []
    any_failure = False

    # 1. Global events (gap, fishing, loitering, encounter)
    for event_type, dataset in WORLD_DATASETS.items():
        started_at = datetime.now(timezone.utc).isoformat()
        target_path = (
            f"{BRONZE_ROOT}/gfw/events/{event_type}/load_type=incremental/"
            f"ingest_date={ingest_date}/world_{ym_str}.jsonl"
        )
        try:
            rows, retries = fetch_events_with_retry(
                dataset=dataset,
                start_date=win_start,
                end_date=win_end,
                geometry=None,
                max_rows=MAX_ROWS_PER_CHUNK,
            )
            body = "\n".join(json.dumps(r) for r in rows)
            dbutils.fs.put(target_path, body, True)
            status = "SUCCESS"
            err_msg = None
            print(f"[OK] {event_type} world {ym_str}: {len(rows)} rows -> {target_path}")
        except Exception as exc:
            rows = []
            retries = 5
            status = "FAILED"
            err_msg = str(exc)
            any_failure = True
            print(f"[ERR] {event_type} world {ym_str}: {err_msg}")

        manifest_records.append({
            "run_id": run_id,
            "entity": event_type,
            "dataset": dataset,
            "region": "world",
            "window_start": win_start,
            "window_end": win_end,
            "load_type": "incremental",
            "source": "gfw_v3_events_api",
            "rows_received": len(rows),
            "file_path": target_path,
            "status": status,
            "retry_count": retries,
            "error_message": err_msg,
            "started_at_utc": started_at,
            "finished_at_utc": datetime.now(timezone.utc).isoformat(),
            "source_url": f"{BASE_URL}/events",
            "api_version": "v3",
        })

    # 2. Port visits across 13 major ports
    for port_name, bbox in PORT_BOXES.items():
        geom = bbox_to_polygon(bbox)
        started_at = datetime.now(timezone.utc).isoformat()
        target_path = (
            f"{BRONZE_ROOT}/gfw/events/port_visit/load_type=incremental/"
            f"ingest_date={ingest_date}/{port_name}_{ym_str}.jsonl"
        )
        try:
            rows, retries = fetch_events_with_retry(
                dataset=PORT_VISIT_DATASET,
                start_date=win_start,
                end_date=win_end,
                geometry=geom,
                max_rows=50000,
            )
            body = "\n".join(json.dumps(r) for r in rows)
            dbutils.fs.put(target_path, body, True)
            status = "SUCCESS"
            err_msg = None
            print(f"[OK] port_visit {port_name} {ym_str}: {len(rows)} rows -> {target_path}")
        except Exception as exc:
            rows = []
            retries = 5
            status = "FAILED"
            err_msg = str(exc)
            any_failure = True
            print(f"[ERR] port_visit {port_name} {ym_str}: {err_msg}")

        manifest_records.append({
            "run_id": run_id,
            "entity": "port_visit",
            "dataset": PORT_VISIT_DATASET,
            "region": port_name,
            "window_start": win_start,
            "window_end": win_end,
            "load_type": "incremental",
            "source": "gfw_v3_events_api",
            "rows_received": len(rows),
            "file_path": target_path,
            "status": status,
            "retry_count": retries,
            "error_message": err_msg,
            "started_at_utc": started_at,
            "finished_at_utc": datetime.now(timezone.utc).isoformat(),
            "source_url": f"{BASE_URL}/events",
            "api_version": "v3",
        })

    manifest_path = f"{BRONZE_ROOT}/_manifest/ingestion_manifest/{run_id}.jsonl"
    dbutils.fs.put(manifest_path, "\n".join(json.dumps(r) for r in manifest_records), True)
    print(f"\nManifest written to {manifest_path}")

    # Advance watermark ONLY if all chunks succeeded
    if not any_failure:
        dbutils.fs.put(
            WATERMARK_PATH,
            json.dumps({
                "last_loaded_month": ym_str,
                "window_end": win_end,
                "updated_at_utc": datetime.now(timezone.utc).isoformat(),
                "run_id": run_id,
            }, indent=2),
            True,
        )
        print(f"Watermark advanced to {ym_str} ({win_end})")
    else:
        raise RuntimeError("One or more incremental chunks failed; watermark was NOT advanced.")


if __name__ == "__main__":
    run_incremental_ingestion()
```
